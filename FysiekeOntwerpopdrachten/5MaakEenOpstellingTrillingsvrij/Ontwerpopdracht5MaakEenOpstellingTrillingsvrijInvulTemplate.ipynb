## Opdracht 2b: Opsplitsen taak 2: afleiding ontwerpeis

```python
# Bereken hier je benodigde kantelfrequentie
r = 1/3 # verhouding tussen Am en Av
f = 5 # hz
w = 2*np.pi*f
w0 = np.sqrt(w**2)/(2)
f_kantel = w0/(2*np.pi)
print(f_kantel)
```

```python
# Voeg hier je plot in.
f0 = np.linspace(0.001,10,1000)
r = 1/(1-((w0**2)/((2*np.pi*f0)**2)))

log_f0 = np.log(f0)
log_r = (np.log(np.abs(r)))


plt.figure()
plt.plot(log_f0,log_r,'r-')
plt.xlabel('log van f0 in hz')
plt.ylabel('log van de verhouding tussen Am en Av')
plt.title('log log plot')
plt.show
```

```python
# Bereken hier de minimale Delta-x waarbij je aan de eis van de opdracht voldoet.
g = 9.81 #m/s^2
delta_x = g/(w0**2)
print('De minimale delta_X is {} cm' .format(delta_x*100))
```

## Opdracht 3: Plot de grafiek.

```python
# Voeg hier je plot in. Lees goed de opdracht wat er in de grafiek moet staan!
# Het is ook handig om je grafiek op te slaan, doe dat door met plt.savefig().

dx1 = delta_x
dx2 = delta_x * 2
dx3 = delta_x * 5
m = np.linspace(0,10,100)
C1 = m * g /dx1
C2 = m * g /dx2
C3 = m * g /dx3
plt.figure()
plt.plot(m,C1,label='dx')
plt.plot(m,C2,label='dx*2')
plt.plot(m,C3,label='dx*5')
plt.xlabel('massa in gram')
plt.ylabel('Veerconstante in N*m')
plt.title('massa tegen veerconstante')
plt.legend()
plt.show()
```
